# Self Attention

```python
import torch
import torch.nn as nn
```

```python
class SelfAttention_v1(nn.Module):
    def __init__(self, d_in, d_out):
        super().__init__()
        # Inizializzazione manuale delle matrici dei pesi W_query, W_key, W_value
        self.W_query = nn.Parameter(torch.rand(d_in, d_out))
        self.W_key   = nn.Parameter(torch.rand(d_in, d_out))
        self.W_value = nn.Parameter(torch.rand(d_in, d_out))

    def forward(self, x):
        # 1. Proiezione dell'input per ottenere Keys, Queries e Values
        keys = x @ self.W_key
        queries = x @ self.W_query
        values = x @ self.W_value
        
        # 2. Calcolo dei punteggi di attenzione grezzi (omega)
        attn_scores = queries @ keys.T 
        
        # 3. Scalamento per sqrt(d_k) e normalizzazione tramite Softmax
        attn_weights = torch.softmax(
            attn_scores / keys.shape[-1]**0.5, dim=-1
        )
        
        # 4. Calcolo dei vettori di contesto arricchiti (somma pesata dei Values)
        context_vec = attn_weights @ values
        return context_vec
```
